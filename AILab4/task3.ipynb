```python
# ---------- 8-Puzzle with IDDFS ----------
START = (1, 2, 3, 4, 0, 6, 7, 5, 8)
GOAL  = (1, 2, 3, 4, 5, 6, 7, 8, 0)

# Blank moves: (name, row change, column change), tried in this order
MOVES = [("Up", -1, 0), ("Down", 1, 0), ("Left", 0, -1), ("Right", 0, 1)]


def get_neighbors(state):
    """Return all states reachable by moving the blank once."""
    neighbors = []
    i = state.index(0)
    row, col = divmod(i, 3)
    for name, dr, dc in MOVES:
        r, c = row + dr, col + dc
        if 0 <= r < 3 and 0 <= c < 3:
            j = r * 3 + c
            new = list(state)
            new[i], new[j] = new[j], new[i]
            neighbors.append((name, tuple(new)))
    return neighbors


def dls(state, limit, path, moves, counter):
    """Depth-limited DFS. counter[0] counts nodes expanded."""
    if state == GOAL:
        return True
    if limit == 0:
        return False

    counter[0] += 1                      # node expanded (children generated)
    for name, child in get_neighbors(state):
        if child in path:                # avoid cycles along the current path
            continue
        path.append(child)
        moves.append(name)
        if dls(child, limit - 1, path, moves, counter):
            return True
        path.pop()
        moves.pop()
    return False


def iddfs(start, max_depth=30):
    total_expanded = 0
    for depth in range(max_depth + 1):
        counter = [0]
        path, moves = [start], []
        found = dls(start, depth, path, moves, counter)
        total_expanded += counter[0]
        print(f"Depth limit {depth}: {counter[0]} nodes expanded")
        if found:
            return depth, total_expanded, path, moves
    return None, total_expanded, None, None


def print_board(state):
    for r in range(3):
        print(" ".join(str(x) if x != 0 else "_" for x in state[r*3:r*3+3]))


if __name__ == "__main__":
    depth, expanded, path, moves = iddfs(START)

    print()
    if depth is not None:
        print(f"Goal found at depth: {depth}")
        print(f"Total nodes expanded (all iterations): {expanded}")
        print(f"Moves (blank): {moves}\n")
        for step, s in enumerate(path):
            print(f"Step {step}:")
            print_board(s)
            print()
    else:
        print("Goal not found within max depth.")
```
