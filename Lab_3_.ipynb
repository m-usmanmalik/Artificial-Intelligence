```python
# ---------- Sample tree (adjacency list) ----------
tree = {
    'A': ['B', 'C', 'D'],
    'B': ['E', 'F'],
    'C': ['G'],
    'D': ['H', 'I'],
    'E': ['J', 'K'],
    'F': ['L'],
    'G': ['M'],
    'H': ['N'],
    'I': ['O'],
    'J': [], 'K': [], 'L': [], 'M': [], 'N': [], 'O': []
}


# ---------- BFS with node counter ----------
def bfs(tree, start, goal):
    visited_count = 0
    queue = deque([start])

    while queue:
        node = queue.popleft()
        visited_count += 1              # count each node examined
        if node == goal:
            return True, visited_count
        for child in tree[node]:
            queue.append(child)

    return False, visited_count


# ---------- Depth-Limited DFS with node counter ----------
def dls(tree, node, goal, limit, counter):
    counter[0] += 1                     # count every visit (even repeated ones)
    if node == goal:
        return True
    if limit == 0:
        return False
    for child in tree[node]:
        if dls(tree, child, goal, limit - 1, counter):
            return True
    return False


# ---------- IDDFS with total node counter ----------
def iddfs(tree, start, goal, max_depth=20):
    total_visited = 0
    per_iteration = []

    for depth in range(max_depth + 1):
        counter = [0]                   # nodes visited in this iteration
        found = dls(tree, start, goal, depth, counter)
        total_visited += counter[0]     # accumulate across iterations
        per_iteration.append((depth, counter[0]))
        if found:
            return True, total_visited, per_iteration

    return False, total_visited, per_iteration


# ---------- Main: compare ----------
if __name__ == "__main__":
    start, goal = 'A', 'M'

    bfs_found, bfs_nodes = bfs(tree, start, goal)
    iddfs_found, iddfs_nodes, details = iddfs(tree, start, goal)

    print(f"Start: {start}   Goal: {goal}\n")

    print("IDDFS iteration breakdown:")
    for depth, count in details:
        print(f"  Depth limit {depth}: {count} nodes visited")

    print("\n----- Comparison -----")
    print(f"BFS   found: {bfs_found}, nodes visited: {bfs_nodes}")
    print(f"IDDFS found: {iddfs_found}, total nodes visited: {iddfs_nodes}")

    diff = iddfs_nodes - bfs_nodes
    print(f"\nIDDFS visited {diff} more nodes than BFS "
          f"({iddfs_nodes / bfs_nodes:.2f}x).")
```
